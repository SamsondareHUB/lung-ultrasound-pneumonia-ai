```python
# Synthetic Pipeline Test This notebook tests the training and evaluation pipeline using generated images only. It does not use ultrasound or patient data and cannot measure pneumonia-detection performance.
```

```python
import torch
from torch.utils.data import TensorDataset, DataLoader
from torchvision.models import mobilenet_v2, MobileNet_V2_Weights

torch.manual_seed(42)

# Artificial images and arbitrary labels for a software test only.
x = torch.rand(16, 3, 224, 224)
y = torch.randint(0, 2, (16,))

loader = DataLoader(TensorDataset(x, y), batch_size=4, shuffle=True)

model = mobilenet_v2(weights=MobileNet_V2_Weights.DEFAULT)
model.classifier[1] = torch.nn.Linear(model.classifier[1].in_features, 2)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

images, labels = next(iter(loader))
images, labels = images.to(device), labels.to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
loss_fn = torch.nn.CrossEntropyLoss()

model.train()
optimizer.zero_grad()
logits = model(images)
loss = loss_fn(logits, labels)
loss.backward()
optimizer.step()

print("Device:", device)
print("Batch output shape:", logits.shape)
print("Synthetic loss:", round(loss.item(), 4))
print("Reminder: this is not a pneumonia result.")
```
